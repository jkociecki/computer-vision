Proste progowanie obrazu na składowej V, modelu HSV

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

from PIL import Image
```

```python
img = cv2.imread('kotek.jpg',cv2.IMREAD_COLOR)

img_RGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

Image.fromarray(img_RGB)
```

```python
# zmiana obarzu na obraz w skali szarości
img_Gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

Image.fromarray(img_Gray)
```

```python
_ = plt.hist(img_Gray.flatten(), bins=255)
```

Dokumentację funkcji progującej OpenCV znajdziesz pod linkiem: https://docs.opencv.org/4.5.3/d7/d1b/group__imgproc__misc.html#gae8a4a146d1ca78c626a53577199e9c57

Zwróć uwagę na różne dostępne typy progowania (`Threshold Types`)!

```python
# progowanie obrazu - metoda threshold OpenCV; parametry: obraz, próg, 
# w przypadku cv2.THRESH_BINARY_INV wartości powyżej progu uzyskają wartość trzeciego parametru (255)
(T, img_threshInv) = cv2.threshold(img_Gray, 130, 255, cv2.THRESH_BINARY_INV)

print('Próg: ',T)

Image.fromarray(img_threshInv)
```
