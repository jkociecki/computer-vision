**Histogram:**

```python
# Experiment: effect of the threshold value
thresholds = [40, 60, 80, 100, 120, 140, 160, 180, 200, 220]
fig, axes = plt.subplots(1, len(thresholds), figsize=(20, 6))
for ax, t in zip(axes, thresholds):
    # BINARY_INV: pixels darker than t -> 255, so coins become white
    _, binary = cv2.threshold(gray, t, 255, cv2.THRESH_BINARY_INV)
    ax.imshow(binary, cmap='gray'); ax.set_title(f'threshold = {t}'); ax.axis('off')
plt.show()

# Otsu: threshold chosen automatically (the given value 0 is ignored)
T_otsu, otsu = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
print('Otsu threshold:', T_otsu)
plt.figure(figsize=(6, 8))
plt.imshow(otsu, cmap='gray'); plt.title(f'Otsu, T={T_otsu:.0f}'); plt.axis('off')
plt.show()
```

**Wnioski - wplyw progu:**
- T = 40-80 to za maly prog, zostaja tylko najciemniejsze piksele i z monet sa strzepy, jasna moneta w lewym gornym rogu prawie znika

- T = 200-220 prawie caly obraz bialy, nic z tego nie ma

- Otsu dal T = 127, wynik jest ok ale jasna moneta w lewym gornym rogu ma dziury. Otsu zaklada ze sa 2 klasy a tutaj jasne monety robia jakby trzecia (90-130) wiec prog wyszedl troche za nisko. Recznie dobrany prog ok 145 dalby tu lepszy wynik

Rozwiązanie części 1B:


```python
gray_noisy = cv2.cvtColor(monety_szum, cv2.COLOR_BGR2GRAY)

fig, axes = plt.subplots(1, 3, figsize=(20, 5))
axes[0].hist(gray.ravel(), bins=256, range=(0, 256)); axes[0].set_title('Histogram: clean')
axes[1].hist(gray_noisy.ravel(), bins=256, range=(0, 256)); axes[1].set_title('Histogram: noisy')

T_noisy, otsu_noisy = cv2.threshold(gray_noisy, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
axes[2].imshow(otsu_noisy, cmap='gray'); axes[2].set_title(f'Otsu on noisy image, T={T_noisy:.0f}'); axes[2].axis('off')
plt.show()
```

**Wplyw szumu:**
- na czystym obrazie gorki w histogramie sa waskie i dobrze rozdzielone, a na zaszumionym sie rozlaly i zlaly w jedno, doliny prawie nie ma
- na samych koncach (0 i 255) sa pojedyncze wysokie slupki, wiec chyba jest tez szum typu sol i pieprz
- Otsu daje prawie ten sam prog (125) ale wynik jest duzo gorszy, w tle sa biale kropki a w monetach czarne
- dzieje sie tak bo rozklady na siebie nachodza, czesc pikseli tla przez szum jest ciemniejsza niz prog a czesc pikseli monet jasniejsza. Zadnym jednym progiem sie tego nie naprawi, trzeba najpierw odszumic

```python
# Compare denoising filters with different kernel sizes
ksizes = [3, 5, 9]
filters = {
    'mean':     lambda img, k: cv2.blur(img, (k, k)),
    'gaussian': lambda img, k: cv2.GaussianBlur(img, (k, k), 0),
    'median':   lambda img, k: cv2.medianBlur(img, k),
}

fig, axes = plt.subplots(len(filters), len(ksizes), figsize=(15, 18))
for row, (name, f) in enumerate(filters.items()):
    for col, k in enumerate(ksizes):
        denoised = f(gray_noisy, k)
        T, binary = cv2.threshold(denoised, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
        axes[row, col].imshow(binary, cmap='gray')
        axes[row, col].set_title(f'{name}, k={k}, T={T:.0f}')
        axes[row, col].axis('off')
plt.show()
```

**Wnioski - porownanie filtrow:**
- k = 3 za slabo, w monetach dalej sporo dziur i w tle kropki
- k = 5 juz duzo lepiej dla wszystkich filtrow
- k = 9 monety sa najbardziej jednolite ale monety w srodkowym rzedzie sie sklejaja (przez rozmycie znika szczelina miedzy nimi)
- srednia i mediana wychodza podobnie, gauss przy tym samym k dziala slabiej bo dalsze piksele maja mniejsze wagi
- wybieram mediane bo najlepiej usuwa szum typu sol i pieprz i nie rozmywa tak krawedzi jak srednia
- prog Otsu po filtracji prawie sie nie zmienia (127-128), filtr nie zmienia progu tylko zweza rozklady wiec mniej pikseli jest po zlej stronie

```python
best = cv2.medianBlur(gray_noisy, 9)   # change after looking at the grid above
T_best, binary_best = cv2.threshold(best, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

fig, axes = plt.subplots(1, 3, figsize=(20, 6))
axes[0].imshow(best, cmap='gray'); axes[0].set_title('denoised'); axes[0].axis('off')
axes[1].hist(best.ravel(), bins=256, range=(0, 256)); axes[1].set_title('Histogram after denoising')
axes[2].imshow(binary_best, cmap='gray'); axes[2].set_title(f'Otsu, T={T_best:.0f}'); axes[2].axis('off')
plt.show()
```

**Wynik koncowy 1B:**
- po medianie 9x9 histogram znowu ma wyrazne gorki i doline, wyglada podobnie jak dla czystego obrazu
- Otsu (T = 128) daje czyste monety i czyste tlo
- minus jest taki ze przy k = 9 dwie monety w srodkowym rzedzie sa sklejone, wiec jakbysmy liczyli monety to wyszlaby jedna zamiast dwoch. Do rozdzielania obiektow lepiej byloby wziac mediane k = 5-7 albo dodac jakas morfologie (np erozje)

```python
# Hue histogram of saturated pixels only (background + copper/gold coins)
saturated = s > T_s
plt.figure(figsize=(10, 4))
plt.hist(h[saturated].ravel(), bins=180, range=(0, 180))
plt.title('Hue of saturated pixels'); plt.xlabel('H (0-179)'); plt.ylabel('pixel count')
plt.show()
```

**Odcien nasyconych pikseli:**
- duza gorka przy H = 0-5 i mniejsza przy 176-179 to czerwone tlo
- osobna gorka przy H = 12-25 to miedziane i zlote monety

```python
HUE_LO, HUE_HI = 8, 170   # read from the histogram above

red = (h <= HUE_LO) | (h >= HUE_HI)
background = red & saturated
mask = np.where(background, 0, 255).astype(np.uint8)

plt.figure(figsize=(16, 8)); plt.imshow(mask, cmap='gray'); plt.title('S + H mask'); plt.axis('off'); plt.show()
```

**Maska S + H:**
- tlo = piksel nasycony i czerwony, wszystko inne to moneta
- progi na H (8 i 170) wzielam z doliny w histogramie odcienia
- teraz wszystkie monety, tez kolorowe, sa prawie pelne, zostaly tylko male szczeliny na krawedziach i wytloczeniach

```python
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))
mask_clean = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)          # remove small white specks
mask_clean = cv2.morphologyEx(mask_clean, cv2.MORPH_CLOSE, kernel)   # close small gaps on coin edges

# Fill holes: draw each outer contour as a filled shape, skip tiny blobs
contours, _ = cv2.findContours(mask_clean, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
contours = [c for c in contours if cv2.contourArea(c) > 500]
filled = np.zeros_like(mask_clean)
cv2.drawContours(filled, contours, -1, 255, thickness=cv2.FILLED)

print('Coins found:', len(contours))
plt.figure(figsize=(16, 8)); plt.imshow(filled, cmap='gray'); plt.title('Final mask'); plt.axis('off'); plt.show()
```

**Czyszczenie maski:**
- otwarcie (erozja i potem dylatacja) usuwa male biale plamki z tla
- zamkniecie (dylatacja i potem erozja) zakleja male czarne szczeliny w monetach
- wypelnianie konturow (RETR_EXTERNAL + FILLED) usuwa wszystkie dziury w srodku monet, kontury mniejsze niz 500 px odrzucam bo to smieci
- wyszlo 20 monet czyli tyle ile jest na obrazku (7 + 6 + 7)
- troche poszarpane krawedzie maja monety z ciemnym albo zabkowanym rantem (np 1 zl w srodkowym rzedzie i 5 zl w prawym dolnym rogu)

Przetestuj opracowany powyżej algorytm na poniższym obrazie z wyraźnym zaszumieniem. Co się zmienia i dlaczego? Uzupełnij potok przetwarzania o odszumianie.

```python
monety_kolor_szum = cv2.imread('Monety_kolor_szum.bmp')
plt.subplots(1,1,figsize=(16,8))[1].imshow(monety_kolor_szum[..., ::-1]);
```

```python
def coins_mask(img_bgr):
    hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
    h, s, v = cv2.split(hsv)
    T, _ = cv2.threshold(s, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    background = ((h <= 8) | (h >= 170)) & (s > T)
    mask = np.where(background, 0, 255).astype(np.uint8)
    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    contours = [c for c in contours if cv2.contourArea(c) > 500]
    filled = np.zeros_like(mask)
    cv2.drawContours(filled, contours, -1, 255, thickness=cv2.FILLED)
    return mask, filled, len(contours)

# 1) Same algorithm, no denoising
mask_raw, filled_raw, n_raw = coins_mask(monety_kolor_szum)

# 2) Median blur first, then the same algorithm
denoised = cv2.medianBlur(monety_kolor_szum, 5)
mask_dn, filled_dn, n_dn = coins_mask(denoised)

fig, axes = plt.subplots(2, 2, figsize=(20, 10))
axes[0, 0].imshow(mask_raw, cmap='gray');  axes[0, 0].set_title('noisy: raw mask')
axes[0, 1].imshow(filled_raw, cmap='gray'); axes[0, 1].set_title(f'noisy: filled, coins={n_raw}')
axes[1, 0].imshow(mask_dn, cmap='gray');   axes[1, 0].set_title('median 5: raw mask')
axes[1, 1].imshow(filled_dn, cmap='gray');  axes[1, 1].set_title(f'median 5: filled, coins={n_dn}')
for ax in axes.ravel(): ax.axis('off')
plt.show()
```

**Wnioski - zaszumiony obraz:**
- bez odszumiania wyszlo tylko 12 obiektow zamiast 20
- szum zmienia H i S pojedynczych pikseli, wiec czesc tla przestaje byc "nasycona i czerwona" i trafia do maski monet (biale kropki w tle)
- najgorzej jest na dole obrazka bo tam tlo jest ciemniejsze, a przy malej jasnosci odcien bardzo latwo sie zmienia przez szum
- te kropki lacza sie ze soba i sklejaja monety z dolnego rzedu z tlem w jeden duzy obiekt, stad za malo monet
- po medianie 5x5 (na BGR, przed zamiana na HSV) wyszlo 20 monet, mediana usuwa pojedyncze odstajace piksele a krawedzie monet zostaja ostre
- na dole zostaly jeszcze male kropki ale nie lacza sie z monetami i odrzuca je filtr na pole konturu
- odszumiac trzeba przed HSV i progowaniem bo progi na H i S sa bardzo czule na pojedyncze zle piksele

##### Zad. 3: Tekst

W tym przypadku należy wyodrębnić litery od bardzo nierównomiernie oświetlonego tła. Skorzystaj z binaryzacji metodą Otsu. Zastosuj binaryzację w sposób lokalny (oknem przesuwnym). Jaki wpływ ma rozmiar okna na jakość binaryzacji? Jak przesuwać okno?

_(Pełna punktacja za zadanie wymaga własnoręcznej implementacji okna przesuwnego - samo Otsu możesz wziąć z OpenCV. Użycie zamiast tego_ `cv2.adaptiveThreshold` _liczone będzie jako pół zadania, ale jedynie po demonstracji zrozumienia tej metody.)_

```python
tekst = cv2.imread('Tekst1.bmp')
Image.fromarray(cv2.cvtColor(tekst, cv2.COLOR_BGR2GRAY))
```

Rozwiązanie:

```python
gray_text = cv2.cvtColor(tekst, cv2.COLOR_BGR2GRAY)

T_global, global_otsu = cv2.threshold(gray_text, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
plt.figure(figsize=(10, 10)); plt.imshow(global_otsu, cmap='gray')
plt.title(f'Global Otsu, T={T_global:.0f}'); plt.axis('off'); plt.show()
```

**Globalne Otsu:**
- kartka jest jasna po lewej i bardzo ciemna w prawym dolnym rogu, tlo w ciemnym rogu ma podobna jasnosc co litery w jasnej czesci
- Otsu (T = 89) daje czytelny tekst po lewej ale prawy dolny rog to jedna czarna plama, wszystko ponizej 89 (tlo razem z tekstem) robi sie czarne
- przy nierownym oswietleniu jeden prog na caly obraz nie dziala, trzeba progowac lokalnie

```python
def local_otsu(gray, win, step, min_diff=20):
    H, W = gray.shape
    half = win // 2
    padded = cv2.copyMakeBorder(gray, half, half, half, half, cv2.BORDER_REFLECT)
    thr_map = np.zeros((H, W), dtype=np.float32)

    for y in range(0, H, step):
        for x in range(0, W, step):
            # window of size win centered on the block [y:y+step, x:x+step]
            cy, cx = y + step // 2, x + step // 2
            window = padded[cy:cy + win, cx:cx + win]

            T, _ = cv2.threshold(window, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
            dark, bright = window[window <= T], window[window > T]

            # if Otsu's two classes are barely different, the window is only paper (+ noise)
            if len(dark) == 0 or len(bright) == 0 or bright.mean() - dark.mean() < min_diff:
                T = 0   # everything is background

            # the threshold is applied only to the central block, not the whole window
            thr_map[y:y + step, x:x + step] = T

    # text = darker than local threshold -> black on white
    return np.where(gray < thr_map, 0, 255).astype(np.uint8)
```

**Jak dziala moja funkcja:**
- obraz jest poszerzony o pol okna z kazdej strony (odbicie lustrzane) zeby okna przy krawedziach tez mialy pelny rozmiar
- okno przesuwa sie co step pikseli, dla kazdego bloku step x step licze prog Otsu z wiekszego okna win x win ktore ma ten blok na srodku
- prog przypisuje tylko do srodkowego bloku, bo okno musi byc duze zeby byly w nim i litery i tlo, a prog ma sie zmieniac plynnie
- min_diff: jak w oknie jest sama kartka to Otsu i tak dzieli piksele na 2 klasy (szum papieru). Jesli srednie tych klas roznia sie mniej niz min_diff to uznaje ze to samo tlo

```python
sizes = [25, 75, 151, 401]
fig, axes = plt.subplots(2, 2, figsize=(20, 20))
for ax, w in zip(axes.ravel(), sizes):
    ax.imshow(local_otsu(gray_text, win=w, step=w // 4), cmap='gray')
    ax.set_title(f'window = {w}, step = {w // 4}'); ax.axis('off')
plt.show()
```

**Wnioski - rozmiar okna:**
- win = 25 okno jest wielkosci litery, czesto ma w sobie samo tlo albo kawalek litery i nie ma 2 klas, przez to litery czesciowo znikaja (np "equa!!y !it", "a1a!ab!e")
- win = 75 i 151 dzialaja bardzo dobrze, tekst jest czytelny wszedzie, tez w ciemnym rogu
- win = 401 okno jest za duze, oswietlenie zmienia sie w srodku okna i wraca problem jak przy globalnym Otsu. Na granicy swiatla i cienia sa ciemne kwadratowe plamy i czesc slow znika
- okno musi byc wieksze niz litery i odstepy miedzy nimi, ale mniejsze niz to jak szybko zmienia sie oswietlenie, tutaj najlepiej ok 75-151 px

```python
w = 151
fig, axes = plt.subplots(1, 2, figsize=(20, 10))
axes[0].imshow(local_otsu(gray_text, w, step=w), cmap='gray');      axes[0].set_title('step = window (no overlap)')
axes[1].imshow(local_otsu(gray_text, w, step=w // 4), cmap='gray'); axes[1].set_title('step = window/4 (overlap)')
for ax in axes: ax.axis('off')
plt.show()
```

**Wnioski - przesuwanie okna:**
- step = win (bez nakladania) to obraz jest podzielony na osobne kafelki, prog zmienia sie skokowo na granicach i czesc tekstu jest wyraznie pogrubiona (np "distracting background elements such as")
- step = win/4 (z nakladaniem) prog zmienia sie plynniej i wynik jest rowniejszy
- minus jest taki ze przy win/4 jest ok 16 razy wiecej okien (4 razy wiecej w kazdym wymiarze) wiec liczy sie dluzej

```python
# Effect of the flat-region check: too low -> noise on paper, too high -> dark text erased
fig, axes = plt.subplots(1, 3, figsize=(30, 10))
for ax, m in zip(axes, [0, 10, 20]):
    ax.imshow(local_otsu(gray_text, win=75, step=18, min_diff=m), cmap='gray')
    ax.set_title(f'window = 75, min_diff = {m}'); ax.axis('off')
plt.show()
```
